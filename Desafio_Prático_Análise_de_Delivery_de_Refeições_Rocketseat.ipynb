## Análise de Dados de Delivery de Refeições

### 1. Importações

```python
import pandas as pd
import numpy as np
```

### 2. Carregamento dos Dados

```python
try:
    pedidos_df = pd.read_csv('/content/pedidos.csv')
    print("Arquivo 'pedidos.csv' carregado com sucesso.")
except FileNotFoundError:
    print("Erro: Arquivo 'pedidos.csv' não encontrado. Verifique o caminho.")
    pedidos_df = pd.DataFrame()
```

### 3. Análise Exploratória de Dados (EDA)

```python
print("Primeiras 5 linhas do DataFrame de Pedidos:")
display(pedidos_df.head())

print("Últimas 5 linhas do DataFrame de Pedidos:")
display(pedidos_df.tail())

print("Informações gerais do DataFrame de Pedidos:")
display(pedidos_df.info())

print("Estatísticas descritivas do DataFrame de Pedidos:")
display(pedidos_df.describe())

print(f"Quantidade de registros: {len(pedidos_df)}")
print(f"Quantidade de colunas: {len(pedidos_df.columns)}")
print(f"Tipos de dados por coluna:\n{pedidos_df.dtypes}")
```

### 4. Criação de Novas Colunas (Feature Engineering)

```python
# Calcula a receita de cada item do pedido
pedidos_df['Receita_Item'] = pedidos_df['Quantidade'] * pedidos_df['Preco_Unitario']
print("DataFrame com a nova coluna 'Receita_Item':")
display(pedidos_df.head())
```

### 5. Tratamento de Valores Ausentes

```python
print("Valores nulos antes do tratamento:")
display(pedidos_df.isnull().sum())

# Preenche valores ausentes na coluna 'Quantidade' com a média
if 'Quantidade' in pedidos_df.columns and pedidos_df['Quantidade'].isnull().any():
    media_quantidade = pedidos_df['Quantidade'].mean()
    pedidos_df['Quantidade'] = pedidos_df['Quantidade'].fillna(media_quantidade)
    print(f"Valores ausentes em 'Quantidade' preenchidos com a média: {media_quantidade:.2f}")

# Remove linhas com valores nulos na coluna 'Preco_Unitario'
if 'Preco_Unitario' in pedidos_df.columns and pedidos_df['Preco_Unitario'].isnull().any():
    pedidos_df.dropna(subset=['Preco_Unitario'], inplace=True)
    print("Linhas com valores nulos em 'Preco_Unitario' removidas.")

print("Valores nulos após o tratamento:")
display(pedidos_df.isnull().sum())
```

### 6. Agregações por Item

```python
# Agrupa por 'Item' e calcula quantidade total vendida e receita total
agregacoes_por_item = pedidos_df.groupby('Item').agg(
    Quantidade_Total_Vendida=('Quantidade', 'sum'),
    Receita_Total_Por_Item=('Receita_Item', 'sum')
).reset_index()

print("Agregações por Item:")
display(agregacoes_por_item.head())

# Top 5 itens mais vendidos por quantidade
top_5_quantidade = agregacoes_por_item.sort_values(by='Quantidade_Total_Vendida', ascending=False).head(5)
print("\nTop 5 itens mais vendidos (por quantidade):")
display(top_5_quantidade)

# Top 5 itens que geraram mais receita
top_5_receita = agregacoes_por_item.sort_values(by='Receita_Total_Por_Item', ascending=False).head(5)
print("\nTop 5 itens que geraram mais receita:")
display(top_5_receita)
```

### 7. Análise Temporal

```python
# Converte a coluna 'Data' para o tipo datetime
pedidos_df['Data'] = pd.to_datetime(pedidos_df['Data'])

# Extrai o mês de cada pedido
pedidos_df['Mes'] = pedidos_df['Data'].dt.month

# Calcula a receita total por mês
receita_por_mes = pedidos_df.groupby('Mes')['Receita_Item'].sum().reset_index()
receita_por_mes['Mes'] = receita_por_mes['Mes'].map({1: 'Janeiro', 2: 'Fevereiro', 3: 'Março', 4: 'Abril',
                                                     5: 'Maio', 6: 'Junho', 7: 'Julho', 8: 'Agosto',
                                                     9: 'Setembro', 10: 'Outubro', 11: 'Novembro', 12: 'Dezembro'})

print("Receita total por mês:")
display(receita_por_mes)
```

### 8. Integração de Dados (Merge)

```python
try:
    cardapio_df = pd.read_csv('/content/cardapio.csv')
    print("Arquivo 'cardapio.csv' carregado com sucesso.")
except FileNotFoundError:
    print("Erro: Arquivo 'cardapio.csv' não encontrado. Verifique o caminho.")
    cardapio_df = pd.DataFrame() # Create an empty DataFrame to avoid errors

if not cardapio_df.empty:
    print("Primeiras 5 linhas do DataFrame de Cardápio:")
    display(cardapio_df.head())

    # Faz o merge entre pedidos e cardápio usando a coluna 'Item'
    # Renomear a coluna 'Nome do item' para 'Item' no cardapio_df para o merge funcionar
    if 'Nome do item' in cardapio_df.columns:
        cardapio_df.rename(columns={'Nome do item': 'Item'}, inplace=True)

    pedidos_com_cardapio_df = pd.merge(pedidos_df, cardapio_df[['Item', 'Categoria']], on='Item', how='left')

    print("\nDataFrame de Pedidos após o merge com informações do Cardápio:")
    display(pedidos_com_cardapio_df.head())

    # Calcula a receita total por Categoria
    receita_por_categoria = pedidos_com_cardapio_df.groupby('Categoria')['Receita_Item'].sum().reset_index()

    print("\nReceita total por Categoria:")
    display(receita_por_categoria)

    # Identifica qual categoria gera mais receita
    categoria_mais_receita = receita_por_categoria.sort_values(by='Receita_Item', ascending=False).iloc[0]
    print(f"\nA categoria que gera mais receita é: {categoria_mais_receita['Categoria']} com R$ {categoria_mais_receita['Receita_Item']:.2f}")
else:
    print("Não foi possível realizar o merge, 'cardapio.csv' não foi carregado corretamente.")
```

### 9. Filtros e Consultas

```python
if 'pedidos_com_cardapio_df' in locals() and not pedidos_com_cardapio_df.empty:
    # Filtra pedidos da categoria 'Salgados' com quantidade superior a 10 unidades
    pedidos_salgados_grandes = pedidos_com_cardapio_df[
        (pedidos_com_cardapio_df['Categoria'] == 'Salgados') &
        (pedidos_com_cardapio_df['Quantidade'] > 10)
    ]

    print("Pedidos da categoria 'Salgados' com quantidade superior a 10 unidades:")
    display(pedidos_salgados_grandes)
else:
    print("DataFrame 'pedidos_com_cardapio_df' não disponível para filtro.")
```

### 10. KPIs e Análise Estatística com NumPy

```python
# Calcula os principais indicadores do negócio
if not pedidos_df.empty:
    receita_total = pedidos_df['Receita_Item'].sum()
    total_itens_vendidos = pedidos_df['Quantidade'].sum()
    # Use 'Data' column for unique orders count, as 'Data do pedido' does not exist.
    numero_de_pedidos = pedidos_df['Data'].nunique() # Contagem de datas únicas como proxy para pedidos
    ticket_medio = receita_total / numero_de_pedidos if numero_de_pedidos > 0 else 0

    print(f"Receita Total: R$ {receita_total:.2f}")
    print(f"Total de Itens Vendidos: {total_itens_vendidos:.0f}")
    print(f"Número de Pedidos (proxy por datas únicas): {numero_de_pedidos}")
    print(f"Ticket Médio: R$ {ticket_medio:.2f}")

    # Desafio Extra: Calcular percentis para 'Preco_Unitario' e 'Quantidade'
    if 'Preco_Unitario' in pedidos_df.columns and 'Quantidade' in pedidos_df.columns:
        percentis_preco_unitario = np.percentile(pedidos_df['Preco_Unitario'].dropna(), [25, 50, 75])
        percentis_quantidade = np.percentile(pedidos_df['Quantidade'].dropna(), [25, 50, 75])

        print("\nPercentis para Preco_Unitario (25%, 50%, 75%):")
        print(f"25º Percentil: {percentis_preco_unitario[0]:.2f}")
        print(f"50º Percentil (Mediana): {percentis_preco_unitario[1]:.2f}")
        print(f"75º Percentil: {percentis_preco_unitario[2]:.2f}")

        print("\nPercentis para Quantidade (25%, 50%, 75%):")
        print(f"25º Percentil: {percentis_quantidade[0]:.2f}")
        print(f"50º Percentil (Mediana): {percentis_quantidade[1]:.2f}")
        print(f"75º Percentil: {percentis_quantidade[2]:.2f}")
else:
    print("DataFrame de pedidos não disponível para cálculo de KPIs.")
```
